## Modules, Functions and F-strings


Modules can be imported using the following:
```python

```
and used like so:
```python
alias.function_in_module()
```

Best practise is put import at the first cell.

````python
# Modules
import numpy as np
print(
    np.sin(np.pi / 6),
    np.sqrt(2)
) 
````

````python
def convert_ms_to_kmh(velocity_in_ms):
    '''
    Converts velocity from m/s to km/h
    :param velocity_in_ms: 
    :return: velocity_in_hours
    '''
    return velocity_in_ms * (60 * 60) / 1e3 
    
# We can specify how precisely to print our variables in strings using f-strings
print(convert_ms_to_kmh(5))
print(f'{convert_ms_to_kmh(5):.2f} km/h')  # Fixed Point precision, i.e. to 2 decimal points
print(f'{convert_ms_to_kmh(5):.2e} km/h')  # Scientific Notation to 2 decimals points 
````

## 1) Implement Fizz Buzz

For a given number `x`, if 
- `x` is divisible by 3, print `Fizz`
- `x` is divisible by 5, print `Buzz`
- `x` is divisible by both 3 and 5, print `FizzBuzz`
- `x` does not satisfy any of the conditions above, print the number

What the format should look like  
`1` -> `1`  
`2` -> `2`  
`3` -> `Fizz`  
`4` -> `4`  
`5` -> `Buzz`  
`6` -> `6`  
...  
`15` -> `FizzBuzz`

````python
# x is between [1, 20)
def fizzbuzz(x):
   if x % 5 == 0 and x % 3 == 0:
       return "FizzBuzz"
   elif x % 5 == 0:
       return "Buzz"
   elif x % 3 == 0:
       return "Fizz"
   else:
        return x
        
        
for i, x in enumerate(range(20)):
    print(f'{x} -> {fizzbuzz(x)}')
````
