### Worked Example 4: Find the sum of array

````python
# Arr = [1, 2, 3, ..., 50]
arr = [i for i in range(1, 50)]

# Method 1 
# Using len() 
# Recall sum of 1 + 2 + 3 + ... + n is n(n+1)/2
n = len(arr)
print(f'Sum = {n * (n + 1) / 2}')

# Method 2
# Using [-1] to get the last element
n = arr[-1]
print(f'Sum = {n * (n + 1) / 2}')

# Method 3
# Using a for loop
total = 0
for ele in arr:
    total = total + ele # Scientific Notation to 2 decimals points 

print(f'Sum = {total}')
    
````

````python
def convert_ms_to_kmh(velocity_in_ms):
    '''
    Converts velocity from m/s to km/h
    :param velocity_in_ms: 
    :return: velocity_in_hours
    '''
    return velocity_in_ms * (60 * 60) / 1e3 
    
# We can specify how precisely to print our variables in strings using f-strings
print(convert_ms_to_kmh(5))
print(f'{convert_ms_to_kmh(5):.2f} km/h')  # Fixed Point precision, i.e. to 2 decimal points
print(f'{convert_ms_to_kmh(5):.2e} km/h')  # Scientific Notation to 2 decimals points 
````



### Worked Example 6: Compute Absolute Magnitude
Recall that:
$$
    m - M = 5 \log(d / 10 pc)
$$ 

Create a function that computes absolute magnitude $M$ given distance $d$ and relative magnitude $m$
- Assuming $d$ are in parsec   



````python
def compute_abs_mag(d, m):
    M = m - 5 * np.log(d / 10)
    return M

print(f'{compute_abs_mag(5, 10):.2f}') # Fixed Point precision, i.e. to 2 decimal points
````

## 1) Implement Fizz Buzz

For a given number `x`, if 
- `x` is divisible by 3, print `Fizz`
- `x` is divisible by 5, print `Buzz`
- `x` is divisible by both 3 and 5, print `FizzBuzz`
- `x` does not satisfy any of the conditions above, print the number

What the format should look like  
`1` -> `1`  
`2` -> `2`  
`3` -> `Fizz`  
`4` -> `4`  
`5` -> `Buzz`  
`6` -> `6`  
...  
`15` -> `FizzBuzz`

````python
# x is between [1, 20)
def fizzbuzz(x):
    if x % 3 == 0 and x % 5 == 0:
        return 'FizzBuzz'
    elif x % 3 == 0:
        return 'Fizz'
    elif x % 5 == 0:
        return 'Buzz'
    else:
        return x
        
        
for i, x in enumerate(range(20)):
    print(f'{x} -> {fizzbuzz(x)}')
````

### a) Compute the period of the Sun around the Milky Way (in years)
### b) Estimate how many times has the Sun orbited around the Milky Way since its inception?
### c) Estimate the total mass of the Milky Way (in solar masses) using Newtons Form of Kepler's Third Law and compare it with the accepted value of 1.2 trillion solar masses.
### d) The Sun's rotational velocity should only be around 160 m/s if we measured the visible matter. Explain the discrepancy between observed and measured rotational velocity of the Sun.



You are given the following date:
- Suns distance from centre of MW $R = (8.40 \pm 0.4) kpc$
- Tangential velocity of the Sun $v = (252 \pm 11) km/s$
- Suns age is estimated to be around 4.5 billion years
- $G = 6.67 \times 10^{-11} m^3 kg^{-1} s^{-2}$
- $M_\odot = 2.00 \times 10^{30} kg$
- 1 $pc$ is $3.26 ly$ or $3.09 \times 10^{16}m$

````python
R = 8.33    # kpc 
dR = 0.11   # kpc
v = 252     # kms
dv = 11     # kms
G = 6.67e-11     # m^3 / kg / s^2
M_sun = 2e30     # kg
age_sun = 4.5e9  # years
pc = 3.09e16     # metres
````

````python
# 3) a)
T = 2 * np.pi * (R * 1e3 * pc)/ (v * 1e3)                               # s 
dT = T * np.sqrt((dR / R) ** 2 + (dv / v) ** 2)                         # s
T_in_years = T / (365.25 * 24 * 60 * 60)                                # years        
dT_in_years = T_in_years * np.sqrt((dR / R) ** 2 + (dv / v) ** 2)       # years
print(f'Period is ({T:.2e} +- {dT:.2e}) seconds')
print(f'Period is ({T_in_years / 1e6 :.2f} +- {dT_in_years / 1e6 :.2f}) million years')
````

````python
# 3) b)
print(f'Number of times: ({age_sun / T_in_years:.2f} +- {age_sun / T_in_years * dT_in_years / T_in_years:.2f})')
````

````python
# 3) c)
M = 4 * np.pi ** 2 / G * (R * 1e3 * pc) ** 3 / T ** 2
M_in_solar_masses = M / M_sun
dM_in_solar_mass = M_in_solar_masses * np.sqrt((3 * dR / R) ** 2 + (2 * dT / T) ** 2)
print(f'Mass of MW is ({M_in_solar_masses :.2e} +- {dM_in_solar_mass :.2e}) solar masses')
````

Note that according to Einsteins Shell Theorem, the mass estimates given by Kepler's Third Law only encapsulate mass within the Sun's orbit (i.e. does not account for the mass with an orbit greater than the Sun)
- Suns orbit is slightly elliptical (`e = 0.07`), meaning it should have minimal effect on the result  
- Inclusion of the Dark Matter Halo (which extend much further than the spiral arms)  
- Diameter of MW is ~32 kpc 

The inclusion of dark matter halo contributes the rotational velocity despite not being seen.
